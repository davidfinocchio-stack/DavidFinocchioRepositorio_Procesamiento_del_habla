2) Crear el chatbot utilizando TFIDF y similitud del coseno. (1 punto)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from numpy import dot
from numpy.linalg import norm
```

```python
# Función para calcular la similitud del coseno
def cosine_similarity(vec1, vec2):
    return dot(vec1, vec2) / (norm(vec1) * norm(vec2))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Obtener preguntas y respuestas
preguntas = df["preguntas"].tolist()
respuestas = df["respuestas"].tolist()

# Vectorización TF-IDF
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(preguntas)

# Función del chatbot TF-IDF
def chatbot_tfidf(consulta_usuario):
    consulta_vec = vectorizer.transform([consulta_usuario])
    similitudes = cosine_similarity(consulta_vec, tfidf_matrix).flatten()
    indice_mejor_coincidencia = similitudes.argmax()
    return respuestas[indice_mejor_coincidencia]
```

### 3) Crear otro chatbot utilizando embeddings. Indique cuál embedding (1 punto) pre-entrenado eligió.

RTA: En este caso tras preguntar cuál sería el módelo óptimo para trabajar con embeddings utilice gensim, en especifíco "es_core_news_lg" ya que funciona mejor con el español y se adapta más a los chatbots.




```python
# Instalar spaCy
!pip install spacy

# Descargar modelo en español
!python -m spacy download es_core_news_lg

# Importar librerías
import spacy
import numpy as np

from sklearn.metrics.pairwise import cosine_similarity

#Cargamos el modelo
nlp = spacy.load("es_core_news_lg")


preguntas = df["preguntas"].tolist()
respuestas = df["respuestas"].tolist()


#Función para obtener embeddings
def obtener_embedding(texto):

    doc = nlp(texto)

    return doc.vector

#Crear embeddings
embeddings_preguntas = np.array(
    [obtener_embedding(p) for p in preguntas]
)

def chatbot_embeddings(consulta_usuario):

    # Embedding consulta
    embedding_consulta = obtener_embedding(consulta_usuario)

    # Similitud coseno
    similitudes = cosine_similarity(
        [embedding_consulta],
        embeddings_preguntas
    )

    # Mejor coincidencia
    indice = similitudes.argmax()

    # Retornar respuesta
    return respuestas[indice]
```

c) Implemente una clase ChatBot usando lo elegido en b).

En esta consigna se implementa el chatbot utilizando el dataset del TP4 como base de conocimiento. Se construyen dos sistemas de recuperación: uno basado en TF-IDF y otro basado en embeddings con MiniLM. Ambos permiten buscar la respuesta más similar a la consulta del usuario mediante similitud coseno.

Además, se integra un modelo de lenguaje FLAN-T5 Small que se utiliza únicamente para reformular la respuesta generada por el sistema de embeddings, mejorando su redacción sin cambiar el contenido.

```python
embeddings_kb = modelo_embed_1.encode(
    df["preguntas"].tolist(),
    normalize_embeddings=True
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df["preguntas"])
```

```python
def chatbot_tfidf(pregunta):
    vec = vectorizer.transform([pregunta])
    sims = cosine_similarity(vec, tfidf_matrix).flatten()
    idx = sims.argmax()
    return df["respuestas"][idx]
```

```python
def chatbot_minilm(pregunta):
    emb = modelo_embed_1.encode([pregunta], normalize_embeddings=True)
    sims = cosine_similarity(emb, embeddings_kb)[0]
    idx = sims.argmax()
    return df["respuestas"][idx]
```

```python
def chatbot_llm(pregunta):
    respuesta_base = chatbot_minilm(pregunta)

    prompt = f"""
You are an assistant at a sports club in charge of answering queries.
Respond only in spanish without adding extra information.

Respuesta:
{respuesta_base}
"""

    inputs = tokenizer(prompt, return_tensors="pt", truncation=True)
    outputs = llm.generate(**inputs, max_new_tokens=60)

    return tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
df_res["acierto"] = (df_res["respuesta"] == df_res["esperada"]).astype(int)

resumen = df_res.groupby("modelo")["acierto"].mean().reset_index()
resumen["accuracy"] = resumen["acierto"] * 100

resumen
```

```python
def sim(a, b):
    emb = modelo_embed_1.encode([a, b], normalize_embeddings=True)
    return cosine_similarity([emb[0]], [emb[1]])[0][0]

df_res["relevancy"] = df_res.apply(
    lambda x: sim(x["respuesta"], x["esperada"]),
    axis=1
)
```

```python
final = df_res.groupby("modelo").agg({
    "acierto": "mean",
    "relevancy": "mean"
}).reset_index()

final["accuracy"] = final["acierto"] * 100
final
```
